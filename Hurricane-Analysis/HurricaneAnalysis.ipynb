   Write a function that counts how often each area is listed as an affected area of a hurricane. Store and return the results in a dictionary where the keys are the affected areas and the values are counts of how many times the areas were affected.

   Test your function on your `hurricane` dictionary.

```python
def count_of_affected_areas_of_a_hurricane_fn(hurricane):
    affected_ares_count = {}
    for key, values in hurricane.items():
        for val in values["Areas Affected"]:
            if val in affected_ares_count:
                affected_ares_count[val] += 1
            else:
                affected_ares_count[val] = 1

    return affected_ares_count
affected_areas_hurricane = count_of_affected_areas_of_a_hurricane_fn(consolidated_dict_of_all_huricane_fn(names, months, years, max_sustained_winds, areas_affected, updated_damages_fn(damages, conversion), deaths))
```


7. Just as hurricanes are rated by their windspeed, you want to try rating hurricanes based on other metrics.

   Write a function that rates hurricanes on a mortality scale according to the following ratings, where the key is the rating and the value is the upper bound of deaths for that rating.

   `mortality_scale` `=` `{0: 0,`
                   `1: 100,`
                   `2: 500,`
                   `3: 1000,`
                   `4: 10000}`

   For example, a hurricane with a 1 mortality rating would have resulted in greater than 0 but less than or equal to 100 deaths. A hurricane with a 5 mortality rating would have resulted in greater than 10000 deaths.

   Store the hurricanes in a new dictionary where the keys are mortality ratings and the values are lists containing a dictionary for each hurricane that falls into that mortality rating.

   Test your function on your hurricane dictionary

```python
mortality_scale = {0: 0,
                   1: 100,
                   2: 500,
                   3: 1000,
                   4: 10000}


def mortality_rating_of_hurricane_fn(hurricane, mortality_scale):
    hurricanes_with_mortality_rating = {}
    for rating in mortality_scale:
        hurricanes_with_mortality_rating[rating] = []

    for keys, values in hurricane.items():
        if values["Deaths"] == mortality_scale[0]:
            hurricanes_with_mortality_rating[0].append(values)
        elif values["Deaths"] > mortality_scale[0] and values["Deaths"] <= mortality_scale[1]:
            hurricanes_with_mortality_rating[1].append(values)
        elif values["Deaths"] > mortality_scale[1] and values["Deaths"] <= mortality_scale[2]:
            hurricanes_with_mortality_rating[2].append(values)
        elif values["Deaths"] > mortality_scale[2] and values["Deaths"] <= mortality_scale[3]:
            hurricanes_with_mortality_rating[3].append(values)
        elif values["Deaths"] > mortality_scale[3] and values["Deaths"] <= mortality_scale[4]:
            hurricanes_with_mortality_rating[4].append(values)
            
    return hurricanes_with_mortality_rating

print(mortality_rating_of_hurricane_fn(hurricanes, mortality_scale))
```

```python

def hurricane_rating_on_damage_scale_fn(hurricane):
    #damage scale dictionary
    damage_scale = {0: 0,
                1: 100000000,
                2: 1000000000,
                3: 10000000000,
                4: 50000000000}

    #creating hurricanes_with_damage_rating dictionary with 0-5 rating list
    hurricanes_with_damage_rating = {0:[], 1:[], 2:[], 3:[], 4:[], 5:[]}
    
    for keys, values in hurricane.items():
        if values["Damage"] == "Damages not recorded":
            hurricanes_with_damage_rating[0].append(values)
        elif values["Damage"] == damage_scale[0]:
                hurricanes_with_damage_rating[0].append(values)
        elif values["Damage"] > damage_scale[0] and values["Damage"] <= damage_scale[1]:
            hurricanes_with_damage_rating[1].append(values)
        elif values["Damage"] > damage_scale[1] and values["Damage"] <= damage_scale[2]:
            hurricanes_with_damage_rating[2].append(values)
        elif values["Damage"] > damage_scale[2] and values["Damage"] <= damage_scale[3]:
            hurricanes_with_damage_rating[3].append(values)
        elif values["Damage"] > damage_scale[3] and values["Damage"] <= damage_scale[4]:
            hurricanes_with_damage_rating[4].append(values)
        elif values["Damage"] > damage_scale[4]:
            hurricanes_with_damage_rating[5].append(values)
        
                
    return hurricanes_with_damage_rating
    
print(hurricane_rating_on_damage_scale_fn(hurricanes))
```
